#Reshaping, stacking, squeezing, unsqueezing tesnsors

* Reshaping- reshapes an input tensor to a defined shape
* view- return a view of an inout tensor of certain shape but keep the same memory as the original tensors
* stacking- combines multiple tensors on top of each other(vstack) or side by side(hstack)
* squeez- removes all '1' dimensions from a tensor
* unsqueez - add a '1' dimensions to a targer tensor
* permute - return a view of the input with dimensions permuted (swapped) in a certain way

```python
#create a tensor
import torch
x=torch.arange(1.,9.)
x,x.shape
```

```python
x_reshaped =x.reshape(1,8) #added a single dimension
x_reshaped,x_reshaped.shape
```

```python
x_reshaped=x.reshape(8,1)
x_reshaped,x_reshaped.shape
```

```python
x_reshaped=x.reshape(2,4)
x_reshaped,x_reshaped.shape
```

```python
#change the view
z=x.view(1,8)

z,z.shape,x
```

```python
 #stck tensors on top of each other
x_stacked=torch.stack([x,x,x,x],dim=0)
x_stacked,x_stacked.shape,x_stacked.ndim
```

#Reproducibility

In short how a neural networks learns:

'start with random numbers -> tesnor operation -> update random nubers to try and make them better representations of the data -> again -> again ->......o


To reduce the randomness in neural network and PyTorch comes the concept of **random seed**

Essentially what the random seed does is "flavour" the randomness


```python
import torch

#create two tensors

random_tensor_a=torch.rand(3,4)
random_tensor_b=torch.rand(3,4)

print(random_tensor_a)
print(random_tensor_b)
print(random_tensor_a==random_tensor_b)
```

```python
 #lets make some random but reproducible tensors
import torch
#set the random seed
RANDOM_SEED=42
torch.manual_seed(RANDOM_SEED)
random_tensor_c=torch.rand(3,4)
torch.manual_seed(RANDOM_SEED)
random_tensor_d=torch.rand(3,4)

print(random_tensor_c)
print(random_tensor_d)
print(random_tensor_c==random_tensor_d)




```
